```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

import yfinance as yf
import datetime as dt
```

```python
mkt_index = 'SPY'
equ_index = 'RSP'

end = dt.datetime.today()
start = end - dt.timedelta(days=365*10)

data = yf.download([mkt_index, equ_index], start=start, end=end, auto_adjust=False, threads=True, group_by='ticker')

mkt_weight = data[mkt_index]
equ_weight = data[equ_index]

mkt_weight['log_return'] = np.log(mkt_weight['Close'] / mkt_weight['Close'].shift(1))
equ_weight['log_return'] = np.log(equ_weight['Close'] / equ_weight['Close'].shift(1))

print(f"Market Weight ({mkt_index}) Log Return Standard Deviation:")
print(round(mkt_weight['log_return'].std()*np.sqrt(252)*100, 2),"%") # Annualized Standard Deviation

print(f"\nEqual Weight ({equ_index}) Log Return Standard Deviation:")
print(round(equ_weight['log_return'].std()*np.sqrt(252)*100, 2),"%") # Annualized Standard Deviation
```

```python
# Rolling 1-year (252 trading days) annualized standard deviation comparison
window = 252*1

rolling_vol = pd.DataFrame({
    f"{mkt_index}_rolling_std": mkt_weight["log_return"].rolling(window=window, min_periods=window).std() * np.sqrt(252) * 100,
    f"{equ_index}_rolling_std": equ_weight["log_return"].rolling(window=window, min_periods=window).std() * np.sqrt(252) * 100
})

# Difference: positive means market-weighted is more volatile
rolling_vol["diff_mkt_minus_equ"] = (
    rolling_vol[f"{mkt_index}_rolling_std"] - rolling_vol[f"{equ_index}_rolling_std"]
)

print("Latest rolling 1-year annualized volatility comparison (%):")
print(rolling_vol.dropna().tail(1).T)

pct_equ_higher = (rolling_vol["diff_mkt_minus_equ"] < 0).mean() * 100
print(f"\n{equ_index} had a higher rolling volatility than {mkt_index} in {pct_equ_higher:.2f}% of available windows.")

ax = rolling_vol[[f"{mkt_index}_rolling_std", f"{equ_index}_rolling_std"]].plot(figsize=(12, 5))
ax.set_title(f"Rolling 1-Year Annualized Volatility: {mkt_index} vs {equ_index}")
ax.set_ylabel("Volatility (%)")
ax.set_xlabel("Date")
ax.grid(True, alpha=0.3)
plt.show()
```
